```python
# Read IMDb directly from the URL, so nothing is saved to disk
# The file has about 13M rows, so read in chunks and keep only movies
chunks = pd.read_csv(
    "https://datasets.imdbws.com/title.basics.tsv.gz",
    sep="\t",
    na_values = r"\N",
    quoting = 3,
    chunksize = 1_000_000,
)
imdb_movies = pd.concat(chunk[chunk.titleType == "movie"] for chunk in chunks)

imdb_movies.head()
```
