```python
import kagglehub
import shutil
from pathlib import Path

path = kagglehub.dataset_download("grouplens/movielens-latest-small")

destination = Path("../data/raw/movielens")
shutil.copytree(path, destination, dirs_exist_ok=True)

print("Dataset saved to:", destination)
```

```python
import pandas as pd

ratings_df = pd.read_csv("../data/raw/movielens/ratings.csv")
movies_df = pd.read_csv("../data/raw/movielens/movies.csv")

display(ratings_df.head())
display(movies_df.head())
```

```python
print("Ratings shape:", ratings_df.shape)
print("Movies shape:", movies_df.shape)

display(ratings_df.info())
display(movies_df.info())

display(ratings_df.isna().sum())
display(movies_df.isna().sum())
```

```python
print("Duplicate ratings:", ratings_df.duplicated().sum())

print(
    "Duplicate user-movie pairs:",
    ratings_df.duplicated(subset=["userId", "movieId"]).sum()
)

print(
    "Duplicate movie IDs:",
    movies_df["movieId"].duplicated().sum()
)
```

```python
missing_movies = ~ratings_df["movieId"].isin(movies_df["movieId"])

print("Ratings without a matching movie:", missing_movies.sum())
```

```python
from pathlib import Path

processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

ratings_df.to_csv(processed_dir / "movielens_ratings_cleaned.csv", index=False)
movies_df.to_csv(processed_dir / "movielens_movies_cleaned.csv", index=False)
```
