### Dataset Collection

#### Set up

```python
from pathlib import Path

import kagglehub
import pandas as pd

RAW_DATA_DIRECTORY = Path("../data/raw")
RAW_DATA_DIRECTORY.mkdir(parents=True, exist_ok=True)
```

#### Oscars data

Save the oscar data

```python
oscars_dataset_path = kagglehub.dataset_download("unanimad/the-oscar-award")

oscars = pd.read_csv(
    Path(oscars_dataset_path) / "full_data.csv",
    sep="\t",
    quoting=3,
)

oscars.to_csv(RAW_DATA_DIRECTORY / "oscars.csv", index=False)
print(f"Saved {len(oscars):,} Oscar records")
oscars.head()
```

#### IMDb movies

Save IMDb movies data

```python
imdb_source_url = "https://datasets.imdbws.com/title.basics.tsv.gz"

imdb_chunks = pd.read_csv(
    imdb_source_url,
    sep="\t",
    na_values=r"\N",
    quoting=3,
    chunksize=1_000_000,
)

imdb_movies = pd.concat(
    (chunk.loc[chunk["titleType"] == "movie"] for chunk in imdb_chunks),
    ignore_index=True,
)

imdb_movies.to_csv(RAW_DATA_DIRECTORY / "imdb_movies.csv", index=False)
print(f"Saved {len(imdb_movies):,} IMDb movie records")
imdb_movies.head()
```

```python
imdb_ratings_url = "https://datasets.imdbws.com/title.ratings.tsv.gz"

imdb_ratings = pd.read_csv(
    imdb_ratings_url,
    sep="\t",
    na_values=r"\N",
    quoting=3,
)

imdb_ratings.to_csv(RAW_DATA_DIRECTORY / "imdb_ratings.csv", index=False)
print(f"Saved {len(imdb_ratings):,} IMDb rating records")
imdb_ratings.head()
```
